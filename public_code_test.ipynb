```python
from prompt_engineering.bug_patch import apply_patch
from prompt_engineering.utils import MutationType
```

```python
data = "public_testset"
mutation = MutationType.BUFLEN

edits_import = f"edits.{data}.{mutation}"
submissions_path = f"starter/proxylab-handout/submissions/"
```

```python
# take edits from edits/* and apply to code for each file
import importlib

edits_module = importlib.import_module(edits_import)
edits = edits_module.edits

failed = []

for entry in edits:
    filename = entry["id"]
    e = entry["edits"]
    try:
        apply_patch(submissions_path + filename, e)
    except Exception as ex:
        print(f"Could not apply edits to file {filename}: {ex}")
        failed.append(filename)

print(f"\nDone. {len(failed)} file(s) failed: {failed if failed else 'none'}")
```
